```python
import pdfplumber
from docx import Document

def extract_text(path):
    if path.lower().endswith(".pdf"):
        text = ""
        with pdfplumber.open(path) as pdf:
            for page in pdf.pages:
                text += (page.extract_text() or "") + "\n"
        return text
    elif path.lower().endswith(".docx"):
        doc = Document(path)
        text = "\n".join(p.text for p in doc.paragraphs)                          # открыть документ и склеить абзацы
        return text
    else:
        raise ValueError(f"Неподдерживаемый формат: {path}")
```

```python
print(len(extract_text("../data/resumes/Мадияров Мади cv (1).pdf")))
print(len(extract_text("../data/resumes/Мадияров Мади cv (1).docx")))
```
